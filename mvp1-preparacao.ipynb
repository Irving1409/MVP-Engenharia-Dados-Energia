# MVP de Engenharia de Dados
## Notebook 1 — Preparação do Ambiente

### Objetivo deste notebook

Este notebook é responsável pela preparação do ambiente que será utilizado durante a construção do pipeline de dados.

Nesta etapa serão definidas as estruturas necessárias para organizar e armazenar os dados ao longo do projeto, permitindo a separação lógica entre os diferentes estágios de processamento.

A organização do pipeline seguirá os princípios da Arquitetura Medalhão:

- **Bronze:** dados preservados próximos ao formato de origem.
- **Silver:** dados tratados, padronizados e validados.
- **Gold:** dados modelados e preparados para responder às perguntas de negócio.

Essa separação facilita a rastreabilidade dos dados e torna o processo de transformação mais organizado e reproduzível.


```python
spark.sql("SHOW CATALOGS").show()
```

### Organização do ambiente

Após a verificação dos catálogos disponíveis no ambiente Databricks, foi identificado o catálogo `workspace` como adequado para armazenar os objetos desenvolvidos neste MVP.

Para organizar os diferentes estágios do pipeline, serão criados três schemas dentro desse catálogo:

- `energia_bronze`: armazenamento dos dados provenientes da fonte, mantendo os dados próximos ao formato de origem e preservando seus valores para as etapas posteriores de tratamento.
- `energia_silver`: armazenamento dos dados após os processos de limpeza, padronização e validação.
- `energia_gold`: armazenamento dos dados modelados e preparados para as análises de negócio.

Essa organização permite separar claramente as responsabilidades de cada camada e facilita a rastreabilidade das transformações realizadas durante o pipeline.

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.energia_bronze")
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.energia_silver")
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.energia_gold")

print("Schemas Bronze, Silver e Gold preparados com sucesso.")
```

```python
spark.sql("SHOW SCHEMAS IN workspace").show()
```

## Resultado da preparação

O ambiente do MVP foi preparado com sucesso no Databricks.

- `workspace.energia_gold`

A existência dos schemas foi validada após a criação, garantindo que a estrutura necessária para as próximas etapas do pipeline está disponível.
