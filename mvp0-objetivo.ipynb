### Sobre este notebook

Este notebook apresenta o contexto de negócio, o problema, o objetivo e as perguntas de negócio que irão orientar o desenvolvimento do MVP.

As definições estabelecidas nesta etapa servirão como referência para a coleta, transformação, modelagem e análise dos dados realizadas nos notebooks seguintes.


## 1. Contexto de Negócio

A energia elétrica é um recurso essencial para o funcionamento da sociedade e para o desenvolvimento econômico, sendo utilizada nos setores residencial, comercial, industrial e de serviços. No Brasil, o Sistema Interligado Nacional (SIN) integra diferentes regiões do país e permite a operação coordenada de diversas fontes de geração de energia elétrica.

O Sistema Interligado Nacional conta com diferentes fontes de geração de energia elétrica, entre elas hidrelétrica, térmica, eólica e solar. Ao mesmo tempo, a carga de energia elétrica apresenta variações ao longo do tempo e entre os diferentes subsistemas que compõem o SIN. Compreender o comportamento da geração e da carga permite analisar como essas diferentes fontes participam do atendimento à demanda do sistema elétrico.

Neste projeto será desenvolvido um pipeline de dados na nuvem utilizando dados públicos disponibilizados pelo Operador Nacional do Sistema Elétrico (ONS). O pipeline terá como finalidade realizar a ingestão, organização, tratamento, modelagem e disponibilização dos dados para análise, permitindo investigar o comportamento da geração e da carga de energia elétrica ao longo do tempo.

A utilização desses dados permite construir uma visão estruturada do comportamento do sistema elétrico ao longo do tempo, criando uma base para análises sobre a demanda de energia e a participação das diferentes fontes de geração nos subsistemas do SIN.


## 2. Problema

O Sistema Interligado Nacional (SIN) reúne diferentes fontes de geração de energia elétrica e atende a uma carga que varia ao longo do tempo e entre os diferentes subsistemas do país. Os dados relacionados à geração e à carga estão disponíveis em fontes públicas, porém precisam ser coletados, organizados, tratados e modelados para que possam ser utilizados de forma estruturada em análises.

Diante desse cenário, o problema abordado neste MVP consiste em estruturar dados históricos do setor elétrico brasileiro de forma que seja possível analisar o comportamento da carga de energia e da geração elétrica, considerando diferentes períodos, subsistemas e fontes de geração.

## 3. Objetivo

O objetivo deste MVP é desenvolver um pipeline de dados em ambiente de nuvem utilizando dados públicos disponibilizados pelo Operador Nacional do Sistema Elétrico (ONS), de forma a estruturar informações históricas sobre geração e carga de energia elétrica no Sistema Interligado Nacional.

Para isso, serão realizadas as etapas de ingestão, armazenamento, tratamento, padronização, verificação da qualidade e modelagem dos dados, preparando-os para responder às perguntas de negócio definidas neste projeto.

A solução será implementada no Databricks seguindo os princípios da Arquitetura Medalhão, com organização dos dados nas camadas Bronze, Silver e Gold. A camada final disponibilizará dados tratados e modelados para realização das análises sobre carga, fontes de geração, evolução temporal e comportamento dos diferentes subsistemas.

## 4. Perguntas de Negócio

A partir do contexto e do objetivo definidos para este MVP, foram estabelecidas as seguintes perguntas de negócio para orientar a construção do pipeline e a análise dos dados:

1. **Como a carga de energia elétrica varia ao longo do tempo nos diferentes subsistemas do Sistema Interligado Nacional?**

2. **Como está distribuída a geração de energia elétrica entre as diferentes fontes de geração, como hidrelétrica, térmica, eólica e solar?**

3. **Como evoluiu a participação das diferentes fontes de geração de energia elétrica ao longo do período analisado?**

4. **Como o comportamento da geração de energia elétrica se relaciona com as variações da carga nos diferentes subsistemas do SIN?**

Essas perguntas servirão como referência para as etapas seguintes do projeto, orientando a preparação, transformação, modelagem e análise dos dados.
